### 1.4.4 Temporal Coverage Analysis

This section investigates the temporal distribution of weather observations.

The analysis focuses on identifying the earliest and latest observation timestamps, the overall time span, and the distribution of records across dates.

This helps determine whether the dataset provides sufficient historical coverage for temporal machine learning tasks.

```python
# Convert last_updated to datetime
df["last_updated"] = pd.to_datetime(df["last_updated"])

# Extract temporal information
earliest_date = df["last_updated"].min()
latest_date = df["last_updated"].max()

print(f"Earliest Observation: {earliest_date}")
print(f"Latest Observation: {latest_date}")
print(f"Time Span: {latest_date - earliest_date}")

# Count unique observation dates
print(f"Unique Dates: {df['last_updated'].dt.date.nunique()}")

# Count unique observation timestamps
print(f"Unique Timestamps: {df['last_updated'].nunique()}")
```

**Temporal Coverage — Key Observations**

The dataset spans approximately 880 days, from May 16, 2024, to October 13, 2026.

It contains 28,420 unique observation timestamps across 170,928 records.

However, the overall temporal coverage does not guarantee consistent daily observations for every geographical location. Further investigation is required to assess the temporal distribution across locations.

```python
# Analyze temporal coverage per location

records_per_location = (
    df.groupby("location_name")["last_updated"]
    .agg(["count", "min", "max", "nunique"])
    .sort_values("count")
)

print("Records per location:")
display(records_per_location.describe())

# Inspect locations with the fewest and most records
print("Locations with the fewest records:")
display(records_per_location.head(10))

print("Locations with the most records:")
display(records_per_location.tail(10))
```

### Temporal Distribution — Key Findings

The dataset contains 268 unique geographical locations, with an average of approximately 638 records per location.

However, the number of records varies significantly across locations:

- Minimum records per location: 1
- Maximum records per location: 880
- Average records per location: approximately 638
- Several locations have nearly complete temporal coverage.
- Other locations contain only a single observation.

This indicates substantial temporal imbalance across geographical locations.

**Data Quality Issue — Location Names**

An unusual location name, `-Kingdom`, was identified during the location-level analysis.

This appears to be a potentially malformed or incomplete location name and may indicate a data quality issue.

Further investigation is required to verify the correct location name and determine whether this record should be corrected or excluded during data cleaning.

**Key Implications for Machine Learning**

- Locations with very few observations may not be suitable for temporal modeling.
- Locations with more complete records may be more appropriate for time-series analysis.
- Location-based filtering or careful sampling may be necessary before model development.
- The temporal consistency of observations must be investigated before selecting a forecasting approach.

Overall, the dataset provides broad geographical coverage, but its temporal distribution is not consistent across all locations.

```python
# Analyze time intervals between observations per location

df_sorted = df.sort_values(
    ["location_name", "last_updated"]
).copy()

df_sorted["time_gap_hours"] = (
    df_sorted.groupby("location_name")["last_updated"]
    .diff()
    .dt.total_seconds() / 3600
)

# Overall time gap statistics
display(
    df_sorted["time_gap_hours"].describe()
)

# Time gap statistics per location
gap_summary = (
    df_sorted.groupby("location_name")["time_gap_hours"]
    .agg(["mean", "median", "min", "max", "count"])
)

display(gap_summary.describe())
```

**Temporal Interval Analysis — Key Findings**

The analysis of time intervals between consecutive observations reveals the following:

- The average time gap across all observations is approximately 26.71 hours.
- The median time gap is 24 hours, indicating that the typical observation interval is approximately one day.
- The minimum time gap is 0 hours, suggesting that some consecutive observations share identical timestamps.
- The maximum time gap reaches 17,010 hours (approximately 709 days), indicating substantial temporal interruptions.

The location-level statistics also reveal considerable variation in temporal consistency:

- The median of location-level median time gaps is 24 hours.
- The average of location-level mean time gaps is approximately 1,267 hours.
- Some locations maintain nearly daily observations, while others experience significantly longer intervals between records.

**Conclusion:**

Although the dataset generally follows a daily observation pattern, temporal consistency varies significantly across geographical locations.

This temporal imbalance must be considered when selecting locations and preparing the dataset for time-series machine learning tasks.

```python
# Investigate irregular time intervals

# Zero or negative time gaps
zero_or_negative_gaps = df_sorted[
    df_sorted["time_gap_hours"] <= 0
]

# Gaps greater than 48 hours
large_gaps = df_sorted[
    df_sorted["time_gap_hours"] > 48
]

print(f"Zero or negative gaps: {len(zero_or_negative_gaps)}")
print(f"Gaps greater than 48 hours: {len(large_gaps)}")

# Inspect the largest time gaps
display(
    large_gaps[
        ["country", "location_name", "last_updated", "time_gap_hours"]
    ]
    .sort_values("time_gap_hours", ascending=False)
    .head(10)
)
```

**Irregular Time Intervals — Key Findings**

The investigation identified significant irregularities in the temporal distribution of weather observations.

The main findings are:

- Only one zero or negative time interval was detected.
- A total of 1,082 time gaps exceeded 48 hours.
- The largest observed time gap reached 17,010 hours, equivalent to approximately 709 days.
- Several geographical locations exhibit substantial interruptions in their observation records.
- The largest gaps indicate that some locations have missing periods spanning a considerable portion of the dataset's overall time range.

**Conclusion:**

Although the dataset generally follows a daily observation pattern, significant temporal gaps exist across multiple locations.

These irregularities may affect the reliability of time-series modeling and must be considered during data preparation and model selection.

Further investigation is required before deciding whether to filter locations or handle irregular observations during preprocessing.

### 1.4.6 Validating Temporal Intervals by Geographical Location

In the previous temporal analysis, observations were grouped using the location name only.

However, different countries may contain locations with identical names, which could lead to incorrect time interval calculations.

To improve the accuracy of the analysis, geographical locations will be identified using a combination of country and location name.

This validation aims to:

- Recalculate time intervals using a composite geographical identifier.
- Reassess zero or negative time gaps.
- Identify gaps exceeding 48 hours.
- Compare the validated results with the previous analysis.

This ensures that temporal observations are grouped according to their actual geographical locations.

```python
# Sort observations by country, location, and timestamp

df_sorted = df.sort_values(
    ["country", "location_name", "last_updated"]
).copy()

# Calculate time gaps using a composite geographical identifier

df_sorted["time_gap_hours"] = (
    df_sorted.groupby(["country", "location_name"])["last_updated"]
    .diff()
    .dt.total_seconds() / 3600
)

# Identify irregular intervals

zero_or_negative_gaps = df_sorted[
    df_sorted["time_gap_hours"] <= 0
]

large_gaps = df_sorted[
    df_sorted["time_gap_hours"] > 48
]

# Display results

print(f"Zero or negative gaps: {len(zero_or_negative_gaps)}")
print(f"Gaps greater than 48 hours: {len(large_gaps)}")

print("\nLargest time gaps:")

display(
    large_gaps[
        [
            "country",
            "location_name",
            "last_updated",
            "time_gap_hours"
        ]
    ]
    .sort_values("time_gap_hours", ascending=False)
    .head(10)
)
```

**Temporal Interval Validation — Key Findings**

The temporal interval analysis was repeated using a composite geographical identifier consisting of country and location name.

This approach provides more accurate grouping of observations and prevents records from different countries with identical location names from being combined.

The validated results revealed:

- Zero or negative time intervals: 1
- Time gaps exceeding 48 hours: 1,191
- Maximum time gap: 17,010 hours (approximately 709 days).

The largest time gaps were observed across multiple geographical locations, including Asuncion (Paraguay), Lome (Togo), Surabaya (Indonesia), and Pristina (Kosovo).

An interesting observation is that several locations share similarly large temporal gaps ending around June 4, 2026. This may indicate a common interruption in data collection or availability, although further investigation is needed to establish its cause.

**Conclusion:**

The dataset exhibits substantial temporal irregularities across geographical locations.

Using a composite geographical identifier improves the reliability of temporal analysis and should be maintained in subsequent data processing stages.

These findings indicate that location-specific temporal validation will be necessary before developing time-series machine learning models.
