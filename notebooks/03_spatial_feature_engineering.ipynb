## 3. Spatial Feature Engineering

This section transforms geographical coordinates into additional spatial features that may help machine learning models capture geographical patterns.

The following features will be created:

- `abs_latitude`: Absolute latitude, representing the angular distance from the Equator.
- `is_northern_hemisphere`: Indicates whether a location is in the Northern Hemisphere.
- `is_eastern_hemisphere`: Indicates whether a location is in the Eastern Hemisphere.

The original latitude and longitude values will be preserved.

The initial prediction target for this project is `temperature_celsius`. The usefulness of each feature will be evaluated during model development.

```python
# Standard Libraries
import sys
from pathlib import Path

# Add Project Root to Python Path
sys.path.insert(0, str(Path.cwd().parent))

# Project Configuration
from config import PROJECT_ROOT, RAW_DATA_PATH

# Data Libraries
import pandas as pd

# Load Raw Dataset
df = pd.read_csv(RAW_DATA_PATH)

print("Dataset shape:", df.shape)
```

### 3.1 Creating Spatial Features

The geographical coordinates are transformed into additional numerical features.

The new features provide information about the absolute latitude and the hemispherical position of each observation.

```python
from src.feature_engineering import create_spatial_features

df_spatial = create_spatial_features(df)

spatial_features = [
    "latitude",
    "longitude",
    "abs_latitude",
    "is_northern_hemisphere",
    "is_eastern_hemisphere"
]

display(df_spatial[spatial_features].head(10))
```

### 3.2 Spatial Features Validation

This section validates the newly created spatial features by checking dataset dimensions, statistical distributions, and missing values.

The validation ensures that the new features were generated correctly without introducing missing values or changing the number of observations.

```python
print("Original shape:", df.shape)
print("New shape:", df_spatial.shape)

print("\nNew spatial features:")

display(
    df_spatial[
        [
            "abs_latitude",
            "is_northern_hemisphere",
            "is_eastern_hemisphere"
        ]
    ].describe()
)

print("\nMissing values in spatial features:")

print(
    df_spatial[
        [
            "abs_latitude",
            "is_northern_hemisphere",
            "is_eastern_hemisphere"
        ]
    ].isnull().sum()
)
```

### Key Findings

The spatial feature engineering process successfully generated three additional features without changing the number of observations.

**Dataset Dimensions:**
- Original dataset: 170,928 rows and 41 columns.
- Updated dataset: 170,928 rows and 44 columns.

**Feature Statistics:**
- `abs_latitude` ranges from 0.1395 to 65.30.
- Approximately 78.27% of observations are located in the Northern Hemisphere.
- Approximately 72.03% of observations are located in the Eastern Hemisphere.

**Data Quality Validation:**
- No missing values were found in the newly created spatial features.
- Absolute latitude values are non-negative.
- Hemisphere indicators contain binary values (0 and 1).

**Conclusion:**

The spatial features were successfully created and passed the initial validation checks. The original geographical coordinates were preserved.

Their contribution to predictive performance will be evaluated during model development.

### 3.3 Saving the Processed Dataset

The dataset containing the engineered spatial features is saved in the processed data directory.

This allows subsequent machine learning experiments to reuse the prepared dataset without repeating the feature engineering process.

```python
PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

OUTPUT_PATH = PROCESSED_DIR / "weather_spatial_features.csv"

df_spatial.to_csv(OUTPUT_PATH, index=False)

print("Processed dataset saved successfully.")
print("Saved path:", OUTPUT_PATH)
print("Dataset shape:", df_spatial.shape)
```
